# Data Cleaning

```python
import pandas as pd
import numpy as np
import joblib 
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('G:\\Ethans\\ML\\Projects\\Final Projects\\Airline_ML\\Data\\Airline1.csv')
```

## Handling Null Values

```python
df.isnull().sum()
```

```python
df[df['Source'].isnull()]
df.fillna({'Source': df['Source'].mode()[0]}, inplace=True)
```

```python
df[df['Destination'].isnull()]
df.fillna({'Destination': df['Destination'].mode()[0]}, inplace=True)
```

```python
df[df['Route'].isnull()]
df.fillna({'Route': df['Route'].mode()[0]}, inplace=True)
```

```python
df[df['Price'].isnull()]
df.fillna({'Price': df['Price'].mean()}, inplace=True)
```

```python
df['Total_Stops'].isnull().sum()
df.fillna({'Total_Stops': df['Total_Stops'].mode()[0]}, inplace=True)
```

```python
df.isnull().sum()
```

## Handling Spelling Mistakes

```python
df['Destination'].value_counts()
```

```python
df['Destination'].replace({'New Delhi': 'Delhi'}, inplace=True)
df['Destination'].value_counts()
```

```python
df.columns
```

```python
df
```

## Removing Irrelevant Columns

```python
# Dropping the columns which are not required for the analysis
#ID: Unique identifier for each flight, not useful for prediction.
#Additional_Info: Contains information about the flight, but not useful for prediction.
#Route: Contains the route of the flight, but not useful for prediction.
df.drop(['ID','Additional_Info','Route'], axis=1, inplace=True)
```

```python
df.info()
```

## Converting categorical Columns to Numerical ones

```python
df['Total_Stops'].unique()
```

```python
df['Total_Stops'].replace({'non-stop': 0, '1 stop': 1, '2 stops': 2, '3 stops': 3, '4 stops': 4}, inplace=True)
df['Total_Stops'] = df['Total_Stops'].astype(int)
```

```python
df['Duration'].unique()
```

```python
df['Duration_Hours_Minutes'] =  0

def convert_duration_to_minutes(duration):
    hours = 0
    minutes = 0
    if 'h' in duration:
        hours = int(duration.split('h')[0])
        duration = duration.split('h')[1]
    if 'm' in duration:
        minutes = int(duration.split('m')[0])
    return (hours * 60) + minutes
df['Duration_Hours_Minutes'] = df['Duration'].apply(convert_duration_to_minutes)
```

```python
df['Duration_Hours_Minutes'].unique()
```

```python
df
```

### Converting Date and Time Columns to be Numerical

```python
df['Day_of_Departure'] = pd.to_datetime(df['Date_of_Journey'], format='mixed').dt.day
df['Month_of_Departure'] = pd.to_datetime(df['Date_of_Journey'],format='mixed').dt.month
df['Year_of_Departure'] = pd.to_datetime(df['Date_of_Journey'],format='mixed').dt.year
df['Dep_Time_Hour'] = pd.to_datetime(df['Dep_Time'], format='%H:%M').dt.hour
df['Dep_Time_Minute'] = pd.to_datetime(df['Dep_Time'], format='%H:%M').dt.minute

df['Arrival_Time_Hour'] = 0
df['Arrival_Time_Minute'] = 0
df['Year_of_Arrival'] = 0
df['Month_of_Arrival'] = 0
df['Day_of_Arrival'] = 0

def extract_arrival_time(row):
    arrival_time_str = row['Arrival_Time']
    if ' ' in arrival_time_str:
        date_str, time_str = arrival_time_str.split(' ')
        row['Arrival_Time_Hour'] = pd.to_datetime(time_str, format='%H:%M').hour
        row['Arrival_Time_Minute'] = pd.to_datetime(time_str, format='%H:%M').minute
        row['Year_of_Arrival'] = pd.to_datetime(date_str, format='mixed').year
        row['Month_of_Arrival'] = pd.to_datetime(date_str, format='mixed').month
        row['Day_of_Arrival'] = pd.to_datetime(date_str, format='mixed').day
    else:
        row['Arrival_Time_Hour'] = pd.to_datetime(arrival_time_str, format='%H:%M').hour
        row['Arrival_Time_Minute'] = pd.to_datetime(arrival_time_str, format='%H:%M').minute
        row['Year_of_Arrival'] = row['Year_of_Departure']
        row['Month_of_Arrival'] = row['Month_of_Departure']
        row['Day_of_Arrival'] = row['Day_of_Departure']
    return row

df = df.apply(extract_arrival_time, axis=1)
```

### Dropping the columns that were transformed

```python
df.drop(['Date_of_Journey', 'Dep_Time', 'Arrival_Time', 'Duration'], axis=1, inplace=True)
```

## Outlier Handling

```python
import seaborn as sns
sns.boxplot(x= cleaned_df['Price'],y= cleaned_df['Airline'])
```

```python
cleaned_df['Airline'].unique()
airline_q1_q3 = {
    'IndiGo':[0.25,0.75],
    'Air India': [0.25,0.75],
    'Jet Airways':[0.25,0.75], 
    'SpiceJet':[0.25,0.75],
    'Multiple carriers':[0.25,0.75], 
    'GoAir':[0.25,0.75], 
    'Vistara':[0.25,0.75], 
    'Air Asia':[0.25,0.75],
    'Vistara Premium economy':[0.25,0.75], 
    'Jet Airways Business':[0.25,0.75],
    'Multiple carriers Premium economy':[0.25,0.75]
}
final_df = cleaned_df.copy()
```

```python
for key in cleaned_df['Airline'].unique():#value in airline_q1_q3.items():
    airlineDataSet = ""
    airlineDataSet = df[df['Airline']==key]

    q1 = airlineDataSet['Price'].quantile(0.25)
    q3 = airlineDataSet['Price'].quantile(0.75)
    iqr = q3-q1
    lower_fence = q1 - (1.5 * iqr)
    upper_fence = q3 + (1.5 * iqr)
    print(f'{key}: Lower Fence = {lower_fence}  Upper Fence = {upper_fence}')
    lowerfence_index = airlineDataSet[airlineDataSet['Price']<lower_fence].index
    upperfence_index = airlineDataSet[airlineDataSet['Price']>upper_fence].index

    final_df.drop(lowerfence_index,axis=0,inplace=True)
    final_df.drop(upperfence_index,axis=0,inplace=True)
```
